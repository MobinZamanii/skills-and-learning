# 11. Difference Between `*` and `@`

This distinction is very important.

### `*` → Element-wise Multiplication

```python
A * B
```

Each element is multiplied by the corresponding element.

### `@` → Matrix Multiplication

```python
A @ B
```

Matrix multiplication is performed according to the rules of linear algebra.

````python
# Create two matrices
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("Element-wise multiplication:")
print(A * B)

print("\nMatrix multiplication:")
print(A @ B)
````

### Manual Matrix Multiplication

For the top-left element:

```text
1×5 + 2×7 = 19
```

For the top-right element:

```text
1×6 + 2×8 = 22
```

For the second row:

```text
3×5 + 4×7 = 43
3×6 + 4×8 = 50
```

Therefore:

```text
[19 22]
[43 50]
```

### Dimension Rule

if

```text
A = m × n
B = n × p
```

then

```text
A @ B = m × p
```
The number of columns in the first matrix must be equal to the number of rows in the second matrix.

# 12. Broadcasting

Broadcasting allows NumPy to perform operations between arrays (or scalars) with compatible shapes.

For example:

````python
# Broadcasting a scalar over an array
a = np.array([10, 20, 30, 40])

result = a + 5

print(result)
````

Conceptually, the value `5` is added to every element:

```text
[10, 20, 30, 40]
+
[ 5,  5,  5,  5]
=
[15, 25, 35, 45]
```

Broadcasting is an important concept in NumPy, Machine Learning, and Deep Learning.

# 15. `axis`

One of the most important NumPy concepts, especially when working with matrices and multidimensional data, is `axis`.

Suppose:

```text
1  2  3
4  5  6
```

If we write:

```python
np.sum(a)
```

All elements are summed.

But:

```python
np.sum(a, axis=0)
```

It returns the sum of each column.

و:

```python
np.sum(a, axis=1)
```

It returns the sum of each row.

In simple terms:

```text
axis=0 → result for each column
axis=1 → result for each row
```

````python
# Create a 2D array
a = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("Total:", np.sum(a))
print("Sum over axis=0:", np.sum(a, axis=0))
print("Sum over axis=1:", np.sum(a, axis=1))
````

The same concept applies to statistical functions:

```python
np.mean(a, axis=0)
np.mean(a, axis=1)

np.max(a, axis=0)
np.max(a, axis=1)
```

This concept is very important when working with tabular data and Machine Learning models.

````python
# Calculate statistics along different axes
a = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("Column means:", np.mean(a, axis=0))
print("Row means   :", np.mean(a, axis=1))
````

# 16. Reshaping with `reshape()`

`reshape()` changes the shape of an array without changing the number of elements.

````python
# Create a one-dimensional array
a = np.array([1, 2, 3, 4, 5, 6])

# Reshape the array into 2 rows and 3 columns
matrix = a.reshape(2, 3)

print(matrix)
print("Shape:", matrix.shape)
````

### `reshape` Rule

The total number of elements must remain unchanged.
For example, if the array contains **12 elements**:
```text
12 = 2 × 6
12 = 3 × 4
12 = 1 × 12
```

So these are valid:

```python
a.reshape(2, 6)
a.reshape(3, 4)
a.reshape(1, 12)
```

But:

```python
a.reshape(5, 3)
```

This is not possible because:

```text
5 × 3 = 15
```

## Using `-1`

NumPy can infer one dimension automatically:

```python
a.reshape(2, -1)
```

If there are 6 elements, NumPy understands that the second dimension must be `3`.

````python
# Let NumPy infer the second dimension
a = np.array([1, 2, 3, 4, 5, 6])

matrix = a.reshape(2, -1)

print(matrix)
print(matrix.shape)
````

# 19. Boolean Indexing

We can select part of an array based on a condition.

````python
# Create an array
a = np.array([10, 20, 30, 40, 50])

# Create a Boolean condition
condition = a > 25

print("Condition:", condition)
print("Filtered :", a[condition])
````

We can also write the condition directly:

```python
a[a > 25]
```

This technique is widely used in Data Science.

````python
# Select values greater than 25
a = np.array([10, 20, 30, 40, 50])

print(a[a > 25])
````

## Combining Multiple Conditions

For NumPy conditions, we use:

```text
&  → AND
|  → OR
~  → NOT
```

operators.

It is good practice to wrap each condition in parentheses.

````python
# Create an array
a = np.array([10, 20, 30, 40, 50])

# Select values between 20 and 40
result = a[(a >= 20) & (a <= 40)]

print(result)
````

# 20. `np.where()`

`where()` is useful for finding positions that satisfy a condition or creating conditional outputs.

````python
# Create an array
a = np.array([10, 20, 30, 40, 50])

# Find indices where the condition is true
indices = np.where(a > 25)

print(indices)
````

We can also use the three-argument form of `where`:

```python
np.where(condition, value_if_true, value_if_false)
```

````python
# Replace values greater than 25 with 1, otherwise use 0
a = np.array([10, 20, 30, 40, 50])

result = np.where(a > 25, 1, 0)

print(result)
````

# 27. Saving and Loading Arrays

NumPy provides dedicated formats for saving arrays.

### `np.save()`

Saves a single array to a `.npy` file.

````python
# Create sample data
a = np.array([10, 20, 30, 40, 50])

# Save the array
np.save("sample_array.npy", a)

# Load the array
loaded = np.load("sample_array.npy")

print(loaded)
````

### `np.savez()`

Saves multiple arrays in a single `.npz` file.

````python
# Create two arrays
x = np.array([1, 2, 3])
y = np.array([10, 20, 30])

# Save multiple arrays
np.savez("sample_data.npz", x=x, y=y)

# Load the data
data = np.load("sample_data.npz")

print("x:", data["x"])
print("y:", data["y"])
````

# 28. Practical Example: Patient Data

Now we combine the main concepts in a simple biological example.

Suppose we have the following data:

```text
Age
Tumor Volume
IDH1
```

Example:

- `Age` → Patient age
- `Tumor Volume` → Tumor volume
- `IDH1` → Encoded IDH1 status

> These values are for educational purposes only and are not real patient data.

````python
# Create a small synthetic clinical dataset
patients = np.array([
    [65, 120, 1],
    [52,  80, 0],
    [71, 150, 1],
    [45,  60, 0],
    [60, 100, 1]
])

print(patients)
print("Shape:", patients.shape)
````

## Selecting Columns

To select the first column:

```python
patients[:, 0]
```

For the second column:

```python
patients[:, 1]
```

````python
# Extract individual features
ages = patients[:, 0]
tumor_volume = patients[:, 1]
idh1 = patients[:, 2]

print("Ages:", ages)
print("Tumor volume:", tumor_volume)
print("IDH1:", idh1)
````

## Calculating Statistics

````python
# Calculate basic statistics
print("Mean age:", np.mean(ages))
print("Maximum tumor volume:", np.max(tumor_volume))
print("Minimum tumor volume:", np.min(tumor_volume))
print("Mean tumor volume:", np.mean(tumor_volume))
````

## Filtering Patients

For example, patients older than 60:

````python
# Select patients older than 60 years
older_patients = patients[patients[:, 0] > 60]

print(older_patients)
````

## Patient with the Largest Tumor Volume

First, find the index of the largest tumor volume.

````python
# Find the patient with the largest tumor volume
index = np.argmax(patients[:, 1])

print("Patient index:", index)
print("Patient data :", patients[index])
````

# 29. Simple Normalization

A common operation in Machine Learning is scaling data.

One simple method is:

```text
z = (x - mean) / std
```

In this example, we apply it to age.

````python
# Standardize age values using z-score normalization
ages = patients[:, 0].astype(np.float64)

mean_age = np.mean(ages)
std_age = np.std(ages)

normalized_ages = (ages - mean_age) / std_age

print("Original ages   :", ages)
print("Normalized ages :", normalized_ages)
````

# 31. NumPy Golden Rules

### 1. Always check `shape`

```python
print(X.shape)
```

Many array-related errors are caused by incorrect dimensions.

### 2. Do not forget the difference between `*` and `@`

```python
A * B   # Element-wise multiplication
A @ B   # Matrix multiplication
```

### 3. Learn `axis` well

```python
axis=0   # Operation across rows → result for columns
axis=1   # Operation across columns → result for rows
```

### 4. Boolean indexing is very important

```python
a[a > 10]
```

### 5. `reshape()` does not change the number of elements

```python
a.reshape(...)
```
It only changes the **shape of the data**.

### 6. NumPy is for numerical computing

When we move to Pandas, Scikit-learn, PyTorch, and Bioinformatics projects, we will constantly work with NumPy arrays and similar structures.

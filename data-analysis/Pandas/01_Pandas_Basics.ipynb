# 1. What is Pandas?

Two fundamental Pandas objects are:

```text
Series     → one-dimensional labeled data
```

# 2. Series

A `Series` is a one-dimensional labeled array.

A simple Series contains:

```text
Index → Label
Value → Data
```

````python
ages = pd.Series([25, 30, 35, 40, 45])

print(ages)
````

````python
ages = pd.Series(
    [25, 30, 35],
    index=["Alice", "Bob", "Charlie"]
)

print(ages)
````

### Accessing Series Values

You can access values by position or label.

```python
s.iloc[0]
s.loc["Alice"]
```

````python
print("First value:", ages.iloc[0])
print("Alice's age:", ages.loc["Alice"])
````

### Important Series Attributes

```python
s.index
s.values
s.dtype
s.shape
s.size
s.name
```

````python
scores = pd.Series(
    [85, 90, 78],
    index=["Alice", "Bob", "Charlie"],
    name="Score"
)

print("Index:", scores.index)
print("Values:", scores.values)
print("dtype:", scores.dtype)
print("shape:", scores.shape)
print("size:", scores.size)
print("name:", scores.name)
````

# 3. DataFrame

A `DataFrame` is the main Pandas data structure for tabular data.

You can think of it as a table:

```text
Rows    → observations / samples
Columns → variables / features
```

````python
df = pd.DataFrame({
    "Patient_ID": ["P001", "P002", "P003"],
    "Age": [55, 67, 49],
    "Gender": ["Male", "Female", "Male"],
    "Survival_Days": [420, 650, 310]
})

print(df)
````

## DataFrame Structure

A DataFrame contains rows, columns, an index, values, and data types.

The index identifies rows, while column names identify variables.

````python
print("Shape:")
print(df.shape)

print("\nColumns:")
print(df.columns)

print("\nIndex:")
print(df.index)

print("\nData types:")
print(df.dtypes)
````

# 4. Inspecting a Dataset

Before analyzing a dataset, always inspect it first.

Important methods:

```python
df.head()
df.tail()
df.info()
df.describe()
df.shape
df.columns
df.dtypes
```

````python
print("First rows:")
print(df.head())

print("\nLast rows:")
print(df.tail())

print("\nDataset information:")
df.info()
````

````python
print("Summary statistics:")
print(df.describe())
````

### `head()` and `tail()`

```python
df.head()
```
shows the first 5 rows by default.

```python
df.tail()
```
shows the last 5 rows by default.

You can specify the number of rows:

```python
df.head(10)
df.tail(3)
```

# 5. Creating DataFrames from Dictionaries

A common way to create a DataFrame is from a dictionary of lists.

Each dictionary key becomes a column.

````python
data = {
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [25, 30, 35, 28],
    "Score": [88, 92, 79, 95]
}

students = pd.DataFrame(data)

print(students)
````

# 6. Selecting Columns

Select one column:

```python
df["Age"]
```

This returns a Series.

Select multiple columns:

```python
df[["Age", "Gender"]]
```

This returns a DataFrame.

````python
print("Single column:")
print(df["Age"])

print("\nMultiple columns:")
print(df[["Age", "Gender"]])
````

# 7. Selecting Rows with `loc`

`loc` is label-based indexing.

General syntax:

```python
df.loc[row_label, column_label]
```

````python
print("Row 0:")
print(df.loc[0])

print("\nAge of row 0:")
print(df.loc[0, "Age"])

print("\nRows 0 and 2:")
print(df.loc[[0, 2]])
````

````python
selected = df.loc[[0, 2], ["Patient_ID", "Age"]]

print(selected)
````

# 8. Selecting Rows with `iloc`

`iloc` is position-based indexing.

```python
df.iloc[row_position, column_position]
```

````python
print("First row:")
print(df.iloc[0])

print("\nFirst row, second column:")
print(df.iloc[0, 1])

print("\nFirst two rows:")
print(df.iloc[:2])

print("\nFirst two columns:")
print(df.iloc[:, :2])
````

### `loc` vs `iloc`

```text
loc  → label-based selection
iloc → position-based selection
```

# 9. Filtering Data

Filtering is one of the most important Pandas operations.

Example:

```python
df[df["Age"] > 60]
```

This returns rows where age is greater than 60.

````python
older_patients = df[df["Age"] > 60]

print(older_patients)
````

## Multiple Conditions

Use:

```text
& → AND
| → OR
~ → NOT
```

As with NumPy, put parentheses around individual conditions.

````python
result = df[
    (df["Age"] > 50) &
    (df["Survival_Days"] > 400)
]

print(result)
````

### String Filtering

We can filter text columns with `.str`.

````python
male_patients = df[df["Gender"].str.lower() == "male"]

print(male_patients)
````

# 10. Adding New Columns

We can create a new column by assigning a value or an expression.

````python
df["Age_Years_Next"] = df["Age"] + 1

print(df)
````

### Creating a Column from Multiple Columns

We can create a simple survival category using a condition.

````python
import numpy as np
````

````python
df["Survival_Group"] = np.where(
    df["Survival_Days"] >= 500,
    "Long",
    "Short"
)

print(df)
````

# 11. Basic Statistics

Pandas provides statistical methods similar to NumPy:

```text
mean()
median()
min()
max()
sum()
std()
count()
```

````python
print("Mean age:", df["Age"].mean())
print("Median age:", df["Age"].median())
print("Minimum age:", df["Age"].min())
print("Maximum age:", df["Age"].max())
print("Mean survival:", df["Survival_Days"].mean())
````

# 12. Sorting Data

Important methods:

```python

df.sort_index()
```

Sort by a column:

```python
df.sort_values("Age")
```

Descending order:


# 13. Missing Values

Real datasets often contain missing values.

Important methods:

```python
df.isna()
df.notna()
df.dropna()
df.fillna()
```

````python
data_missing = pd.DataFrame({
    "Age": [55, np.nan, 67, 49],
    "Score": [80, 90, np.nan, 75],
    "Group": ["A", "B", "A", "B"]
})

print(data_missing)

print("\nMissing values per column:")
print(data_missing.isna().sum())
````

### Dropping Missing Values

```python
data_missing.dropna()
```

removes rows containing missing values. Always inspect how much data will be removed before using it on a real dataset.

````python
print(data_missing.dropna())
````

### Filling Missing Values

For example, fill missing numeric values with the column mean.

````python
filled = data_missing.copy()

filled["Age"] = filled["Age"].fillna(filled["Age"].mean())
filled["Score"] = filled["Score"].fillna(filled["Score"].mean())

print(filled)
````

Useful functions include:

````python
example = pd.DataFrame({
    "Age": ["55", "60", "70"],
    "Date": ["2026-01-01", "2026-02-01", "2026-03-01"]
})

example["Age"] = pd.to_numeric(example["Age"])
example["Date"] = pd.to_datetime(example["Date"])

print(example)
print("\nData types:")
print(example.dtypes)
````

# 16. Deleting Columns and Rows

Delete columns with:


# 17. Duplicate Data

Useful methods:

```python
df.duplicated()
df.drop_duplicates()
```

````python
duplicates = pd.DataFrame({
    "Patient_ID": ["P001", "P002", "P002", "P003"],
    "Age": [55, 60, 60, 48]
})

print("Duplicate mask:")
print(duplicates.duplicated())

print("\nAfter removing duplicates:")
print(duplicates.drop_duplicates())
````

# 18. String Operations

Use the `.str` accessor for vectorized string operations:

```python
.str.lower()
.str.upper()
.str.contains()
.str.replace()
.str.strip()
.str.len()
```

````python
names = pd.Series([" Alice ", "BOB", "Charlie Smith", "David"])

print("Lowercase:")
print(names.str.lower())

print("\nUppercase:")
print(names.str.upper())

print("\nContains 'smith':")
print(names.str.lower().str.contains("smith"))

print("\nStripped:")
print(names.str.strip())
````

# 20. Practical Bioinformatics Example

Suppose we have a small clinical dataset with:

```text
Patient_ID
Age
Gender
Tumor_Type
Gene_Expression
Survival_Days
```

The values are synthetic and are only for learning purposes.

````python
bio = pd.DataFrame({
    "Patient_ID": ["P001", "P002", "P003", "P004", "P005", "P006"],
    "Age": [65, 52, 71, 45, 60, 68],
    "Gender": ["Male", "Female", "Male", "Female", "Male", "Female"],
    "Tumor_Type": ["GBM", "GBM", "GBM", "LGG", "GBM", "LGG"],
    "Gene_Expression": [12.4, 8.7, 15.1, 6.2, 11.8, 9.4],
    "Survival_Days": [320, 540, 210, 900, 470, 780]
})

print(bio)
````

````python
print("Shape:", bio.shape)
print("\nData types:")
print(bio.dtypes)

print("\nSummary of numerical columns:")
print(bio.select_dtypes(include="number").describe())
````

````python
# Select GBM patients
gbm = bio[bio["Tumor_Type"] == "GBM"]
print(gbm)

# Patients older than 60
older = bio[bio["Age"] > 60]
print("\nOlder than 60:")
print(older)

# Patients with high gene expression
high_expression = bio[bio["Gene_Expression"] > 10]
print("\nHigh expression:")
print(high_expression)
````

````python
# Find the patient with the longest survival
idx = bio["Survival_Days"].idxmax()

print("Patient with the longest survival:")
print(bio.loc[idx])
````

````python
# Create a survival group
bio["Survival_Group"] = np.where(
    bio["Survival_Days"] >= 500,
    "Long",
    "Short"
)

print(bio)
````

# 22. Common Mistakes

### Mistake 1 — Confusing `loc` and `iloc`

```text
loc  → labels
iloc → positions
```

### Mistake 2 — Incorrect multiple-condition syntax

Correct:

```python
df[(df["Age"] > 50) & (df["Score"] > 80)]
```

### Mistake 3 — Ignoring missing values

Check:

```python
df.isna().sum()
```

### Mistake 4 — Forgetting `index=False` when exporting CSV

```python
df.to_csv("data.csv", index=False)
```

### Mistake 5 — Performing calculations without checking dtypes

```python
df.dtypes
```

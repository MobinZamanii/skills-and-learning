# 4. String methods for sequence analysis

Useful built-ins and methods include `len()`, `.count()`, `.find()`, `.replace()`, `.upper()`, `.lower()`, `.startswith()`, and `.endswith()`.

# 5. DNA validation

A biological sequence should not automatically be assumed to be valid. For standard DNA, a simplified teaching alphabet is `A`, `T`, `C`, and `G`.

````python
def is_valid_dna(sequence):
    # Define the valid DNA alphabet.
    valid_bases = set("ATCG")

    # Normalize the input so lowercase DNA is accepted.
    sequence = sequence.upper()

    # Every character must belong to the allowed alphabet.
    return bool(sequence) and all(base in valid_bases for base in sequence)

print(is_valid_dna("ATGCGTAC"))
print(is_valid_dna("ATGXC"))
````

### Why use a set?

A set stores unique values and is convenient for membership testing. `all()` returns `True` only when every tested condition is true.

# 6. DNA composition and GC content

A basic sequence statistic is GC content:

`GC% = (G + C) / sequence_length × 100`

This is a descriptive statistic. Its biological interpretation depends on the organism, genomic region, sequence type, and scientific question.

````python
dna = "ATGCGTAC"

g_count = dna.count("G")
c_count = dna.count("C")
gc_percentage = (g_count + c_count) / len(dna) * 100

print("G count:", g_count)
print("C count:", c_count)
print("GC content:", gc_percentage, "%")
````

````python
def gc_content(sequence):
    # Normalize the sequence.
    sequence = sequence.upper()

    # Reject an empty sequence.
    if len(sequence) == 0:
        raise ValueError("Sequence cannot be empty.")

    # Count G and C bases.
    gc_count = sequence.count("G") + sequence.count("C")

    # Convert the fraction to a percentage.
    return gc_count / len(sequence) * 100

print(f"GC content: {gc_content('ATGCGTAC'):.2f}%")
````

# 13. `any()` and `all()`

These functions are especially useful for quality control and validation.

````python
sequence = "ATGCGTAC"
valid_bases = set("ATCG")

# Every base must be valid.
all_valid = all(base in valid_bases for base in sequence)

# At least one G must be present.
contains_g = any(base == "G" for base in sequence)

print("All bases valid:", all_valid)
print("Contains G:", contains_g)
````

# 14. Functions: reusable biological operations

A function should have a clear purpose, well-defined inputs, useful outputs, and appropriate validation.

````python
def nucleotide_counts(sequence):
    # Normalize input.
    sequence = sequence.upper()

    # Return counts in a structured dictionary.
    return {
        "A": sequence.count("A"),
        "T": sequence.count("T"),
        "C": sequence.count("C"),
        "G": sequence.count("G"),
    }

print(nucleotide_counts("ATGCGTACG"))
````

# 17. Transcription and reverse complement

A simplified transcription representation replaces `T` with `U`. Reverse complement uses `A↔T` and `C↔G`. Real biological workflows require attention to strand, annotation, and context.

````python
def transcribe_dna(sequence):
    # Validate the simplified DNA alphabet first.
    if not is_valid_dna(sequence):
        raise ValueError("Input must be a non-empty DNA sequence containing A, T, C, and G.")

    return sequence.upper().replace("T", "U")

print(transcribe_dna("ATGCGTAC"))
````

````python
def reverse_complement(sequence):
    # Validate input before transformation.
    if not is_valid_dna(sequence):
        raise ValueError("Input must be a non-empty DNA sequence containing A, T, C, and G.")

    # Define complementary base substitutions.
    complement_table = str.maketrans({
        "A": "T", "T": "A", "C": "G", "G": "C"
    })

    # Translate the bases and reverse the resulting sequence.
    return sequence.upper().translate(complement_table)[::-1]

print(reverse_complement("ATGC"))
````

# 18. Codons and reading frames

A codon is a group of three nucleotides. Stepping through a sequence by three is a basic programming pattern for sequence analysis.

````python
dna = "ATGCGTACG"

# Read complete groups of three from frame 0.
for i in range(0, len(dna) - 2, 3):
    codon = dna[i:i + 3]
    print(i, codon)
````

# 20. Exception handling

Biological data can be messy: empty sequences, invalid characters, missing metadata, malformed files, and unexpected values. Exceptions allow programs to handle expected problems explicitly.

````python
def safe_gc_content(sequence):
    try:
        return gc_content(sequence)
    except ValueError as error:
        return f"Error: {error}"

print(safe_gc_content("ATGCGTAC"))
print(safe_gc_content(""))
````

# 22. Complete DNA analyzer

We now combine the concepts into a small reusable program. It reports length, nucleotide counts, GC content, RNA, and reverse complement.

````python
def analyze_dna(sequence):
    # Normalize the input.
    sequence = sequence.upper()

    # Validate before performing biological calculations.
    if not is_valid_dna(sequence):
        raise ValueError("Invalid DNA sequence: use A, T, C, and G only.")

    counts = nucleotide_counts(sequence)
    gc = gc_content(sequence)

    return {
        "sequence": sequence,
        "length": len(sequence),
        "A": counts["A"],
        "T": counts["T"],
        "C": counts["C"],
        "G": counts["G"],
        "GC_percent": gc,
        "RNA": transcribe_dna(sequence),
        "reverse_complement": reverse_complement(sequence),
    }

result = analyze_dna("ATGCGTACG")
for key, value in result.items():
    print(f"{key}: {value}")
````

````python
def print_dna_report(sequence):
    # Run the complete analysis.
    result = analyze_dna(sequence)

    print("=" * 40)
    print("DNA SEQUENCE ANALYSIS")
    print("=" * 40)
    print(f"Sequence:              {result['sequence']}")
    print(f"Length:                {result['length']}")
    print(f"A count:               {result['A']}")
    print(f"T count:               {result['T']}")
    print(f"C count:               {result['C']}")
    print(f"G count:               {result['G']}")
    print(f"GC content:            {result['GC_percent']:.2f}%")
    print(f"RNA:                   {result['RNA']}")
    print(f"Reverse complement:    {result['reverse_complement']}")
    print("=" * 40)

print_dna_report("ATGCGTACG")
````

# 23. Batch analysis

Real projects usually analyze many sequences. Here we store records in a dictionary and produce a compact summary.

````python
samples = {
    "sample_01": "ATGCGTAC",
    "sample_02": "GGGCCCAA",
    "sample_03": "ATATATAT",
    "sample_04": "CGCGCGCG",
}

results = []

for sample_id, sequence in samples.items():
    result = analyze_dna(sequence)
    result["sample_id"] = sample_id
    results.append(result)

for result in results:
    print(result["sample_id"], f"length={result['length']}", f"GC={result['GC_percent']:.2f}%")
````

# 24. Data quality and ambiguous bases

The strict validator accepts only `A`, `T`, `C`, and `G`. Real sequencing data can contain ambiguity codes such as `N`, `R`, `Y`, `S`, `W`, `K`, `M`, `B`, `V`, `D`, and `H`. Therefore, validation rules must match the biological data source and scientific question. Do not blindly reject real biological data because it contains a character excluded from a simplified teaching alphabet.

````python
def is_valid_dna_with_n(sequence):
    # Allow N as an ambiguous base in this simplified example.
    allowed_bases = set("ATCGN")
    sequence = sequence.upper()
    return bool(sequence) and all(base in allowed_bases for base in sequence)

print(is_valid_dna_with_n("ATGCNN"))
print(is_valid_dna_with_n("ATGX"))
````

# 25. Mini-project — DNA Quality Control Pipeline

Build a pipeline that reports sample ID, length, validity, GC content when valid, and QC status. The reference implementation is provided below after the task.

````python
samples = {
    "S01": "ATGCGTAC",
    "S02": "GGCCAA",
    "S03": "ATGXCG",
    "S04": "ATAT",
}

minimum_length = 6

for sample_id, sequence in samples.items():
    sequence = sequence.upper()
    valid = is_valid_dna(sequence)
    length = len(sequence)

    print(f"Sample: {sample_id}")
    print(f"  Sequence: {sequence}")
    print(f"  Length: {length}")
    print(f"  Valid DNA: {valid}")

    if valid and length >= minimum_length:
        gc = gc_content(sequence)
        print(f"  GC content: {gc:.2f}%")
        print("  QC status: PASS")
    else:
        print("  QC status: FAIL")

    print("-" * 40)
````

# Matplotlib 3D Visualization — Complete Guide

A practical, parameter-focused guide to 3D visualization with Matplotlib, from basic coordinates to surfaces, contours, camera control, vector fields, voxels, medical-imaging concepts, scientific examples, performance, and projects.

> Use 3D when the third dimension adds meaningful information.

## Learning Goals

- Create 3D axes and understand X/Y/Z.
- Build 3D lines, scatter plots, bars, surfaces, wireframes, contours, vector fields, and voxels.
- Understand `meshgrid()`, colormaps, normalization, colorbars, camera settings, aspect ratio, and projections.
- Apply 3D visualization to clinical, bioinformatics, machine-learning, and Cancer AI examples.
- Export figures and avoid common scientific-visualization mistakes.

## 1. Setup

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # Educational import for Matplotlib 3D support.

plt.rcParams["figure.figsize"] = (8, 6)
```

## 2. The 3D Coordinate System

A 3D point is `(x, y, z)`. X, Y, and Z must have clearly defined scientific meanings. Examples include Age/Tumor Volume/Survival Days, or spatial X/Y/Z coordinates.

```python
fig = plt.figure(figsize=(8, 6))

# 111 = 1 row, 1 column, first subplot.
# projection="3d" creates a 3D Axes object.
ax = fig.add_subplot(111, projection="3d")

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")
plt.show()
```

## 3. Basic 3D Line Plot

```python
t = np.linspace(0, 10, 100)
x = np.cos(t)  # X coordinate
y = np.sin(t)  # Y coordinate
z = t          # Z coordinate

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection="3d")

ax.plot(
    x, y, z,
    linewidth=2,       # Line thickness
    linestyle="-",      # "-" solid, "--" dashed, ":" dotted
    marker="o",         # Marker shape
    markersize=3,       # Marker size
    alpha=0.8,           # 0 = transparent, 1 = opaque
    label="Trajectory"  # Legend label
)

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")
ax.legend()
plt.show()
```

### `plot()` parameters

Important parameters: `x`, `y`, `z`, `color`, `linestyle`, `linewidth`, `marker`, `markersize`, `markerfacecolor`, `markeredgecolor`, `alpha`, `label`, and `zorder`.

## 4. Multiple 3D Lines

```python
t = np.linspace(0, 10, 100)
x1, y1, z1 = np.cos(t), np.sin(t), t
x2, y2, z2 = np.cos(t + np.pi/3), np.sin(t + np.pi/3), t

fig = plt.figure(figsize=(9, 6))
ax = fig.add_subplot(111, projection="3d")
ax.plot(x1, y1, z1, linewidth=2, label="Trajectory 1")
ax.plot(x2, y2, z2, linewidth=2, label="Trajectory 2")
ax.set_xlabel("X"); ax.set_ylabel("Y"); ax.set_zlabel("Z")
ax.legend()
plt.show()
```

## 5. 3D Scatter

```python
rng = np.random.default_rng(42)
x = rng.normal(size=100)
y = rng.normal(size=100)
z = rng.normal(size=100)

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection="3d")
ax.scatter(
    x, y, z,
    s=45,              # Marker size; scalar or array
    alpha=0.75,        # Transparency
    marker="o",       # Marker shape
    depthshade=True    # Depth-based marker shading
)
ax.set_xlabel("X"); ax.set_ylabel("Y"); ax.set_zlabel("Z")
plt.show()
```

### `scatter()` parameters

`xs, ys, zs` = coordinates; `s` = marker size; `c` = numeric values or colors; `marker` = shape; `cmap` = colormap; `norm` = normalization; `alpha` = transparency; `edgecolors` = marker edges; `linewidths` = edge width; `depthshade` = depth shading; `label` = legend label.

```python
rng = np.random.default_rng(7)
x = rng.uniform(0, 10, 100); y = rng.uniform(0, 10, 100); z = rng.uniform(0, 10, 100)
values = x**2 + y**2 + z**2
sizes = 20 + 100 * (values-values.min())/(values.max()-values.min())

fig = plt.figure(figsize=(9, 6)); ax = fig.add_subplot(111, projection="3d")
points = ax.scatter(
    x, y, z,
    s=sizes,           # Array: each point can have a different size
    c=values,          # Fourth variable mapped to color
    cmap="viridis",   # Colormap
    alpha=0.8
)
cbar = fig.colorbar(points, ax=ax, shrink=0.75, aspect=20, pad=0.10)
cbar.set_label("Fourth Variable")
ax.set_xlabel("X"); ax.set_ylabel("Y"); ax.set_zlabel("Z")
plt.show()
```

## 6. Clinical 3D Scatter

```python
rng = np.random.default_rng(21)
age = rng.integers(20, 81, 100)
tumor_volume = rng.uniform(5, 150, 100)
survival_days = rng.uniform(30, 1500, 100)

fig = plt.figure(figsize=(9, 7)); ax = fig.add_subplot(111, projection="3d")
points = ax.scatter(
    age, tumor_volume, survival_days,
    s=45,                    # Patient marker size
    c=survival_days,        # Color encodes survival
    cmap="viridis",         # Colormap
    alpha=0.8
)
fig.colorbar(points, ax=ax, shrink=0.75, pad=0.10, label="Survival Days")
ax.set_xlabel("Age"); ax.set_ylabel("Tumor Volume"); ax.set_zlabel("Survival Days")
ax.set_title("Illustrative Clinical 3D Visualization")
plt.show()
```

## 7. 3D Bar Charts

```python
x = np.array([0,1,2,0,1,2]); y = np.array([0,0,0,1,1,1]); z = np.zeros(6)
dx = np.full(6, 0.6)              # Bar width
dy = np.full(6, 0.6)              # Bar depth
dz = np.array([3,5,2,4,7,6])      # Bar height

fig = plt.figure(figsize=(9,7)); ax = fig.add_subplot(111, projection="3d")
ax.bar3d(x, y, z, dx, dy, dz, alpha=0.8, shade=True)
ax.set_xlabel("X Category"); ax.set_ylabel("Y Category"); ax.set_zlabel("Value")
plt.show()
```

### `bar3d()` parameters

`x, y, z` are bar origins. `dx, dy, dz` are width, depth, and height. Other useful parameters include `alpha`, `shade`, `zsort`, `color`, `edgecolor`, and `linewidth`. 3D bars should be used carefully because perspective can make comparison difficult.

## 8. `meshgrid()` — Foundation of Regular Surfaces

```python
x = np.linspace(-3, 3, 6)
y = np.linspace(-2, 2, 5)

# meshgrid creates every X/Y coordinate combination.
X, Y = np.meshgrid(x, y)
print("x:", x.shape, "y:", y.shape)
print("X:", X.shape, "Y:", Y.shape)

# One Z value for every X/Y location.
Z = np.sin(X) * np.cos(Y)
print("Z:", Z.shape)
```

If x has 6 values and y has 5 values, `meshgrid()` creates a 5×6 coordinate grid. `X`, `Y`, and `Z` therefore describe the same surface locations.

## 9. Surface Plot

```python
x = np.linspace(-4, 4, 120); y = np.linspace(-4, 4, 120)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))

fig = plt.figure(figsize=(9,7)); ax = fig.add_subplot(111, projection="3d")
surface = ax.plot_surface(
    X, Y, Z,
    cmap="viridis",       # Map Z values to colors
    rcount=80,             # Maximum sampled rows
    ccount=80,             # Maximum sampled columns
    linewidth=0,            # Surface mesh line width
    antialiased=True,       # Smooth rendering
    shade=True,             # Shading
    alpha=0.9               # Transparency
)
fig.colorbar(surface, ax=ax, shrink=0.75, pad=0.10, label="Z Value")
ax.set_xlabel("X"); ax.set_ylabel("Y"); ax.set_zlabel("Z")
ax.set_title("3D Surface")
plt.show()
```

### `plot_surface()` parameters

`X,Y,Z` = coordinates; `cmap` = colormap; `norm` = normalization; `vmin/vmax` = color limits; `rstride/cstride` = sampling steps; `rcount/ccount` = maximum rendered rows/columns; `linewidth` = mesh width; `antialiased` = smoothing; `shade` = shading; `alpha` = transparency; `facecolors` = explicit face colors.

## 10. Wireframe

```python
x = np.linspace(-4,4,100); y = np.linspace(-4,4,100); X,Y = np.meshgrid(x,y)
Z = np.sin(np.sqrt(X**2 + Y**2))
fig = plt.figure(figsize=(9,7)); ax = fig.add_subplot(111, projection="3d")
ax.plot_wireframe(
    X,Y,Z,
    rcount=30,       # Render at most this many rows
    ccount=30,       # Render at most this many columns
    linewidth=0.8,   # Wire width
    alpha=0.8        # Transparency
)
ax.set_xlabel("X"); ax.set_ylabel("Y"); ax.set_zlabel("Z")
plt.show()
```

## 11. Irregular Surfaces with `plot_trisurf()`

```python
rng = np.random.default_rng(4)
x = rng.uniform(-4,4,180); y = rng.uniform(-4,4,180)
z = np.sin(np.sqrt(x**2+y**2))

fig = plt.figure(figsize=(9,7)); ax = fig.add_subplot(111, projection="3d")
surface = ax.plot_trisurf(
    x,y,z,
    cmap="viridis",     # Colormap
    linewidth=0.2,       # Triangle edge width
    antialiased=True
)
fig.colorbar(surface, ax=ax, shrink=0.75)
ax.set_xlabel("X"); ax.set_ylabel("Y"); ax.set_zlabel("Z")
plt.show()
```

## 12. 3D Contours and Projection

```python
x = np.linspace(-4,4,100); y = np.linspace(-4,4,100); X,Y = np.meshgrid(x,y)
Z = np.sin(np.sqrt(X**2+Y**2))

fig = plt.figure(figsize=(9,7)); ax = fig.add_subplot(111, projection="3d")
ax.contour3D(X,Y,Z, levels=20, cmap="viridis")  # levels = number of contour levels

# zdir="z" projects a 2D contour along Z; offset controls its Z location.
ax.contour(X,Y,Z, levels=15, zdir="z", offset=-1.5, cmap="viridis")
ax.plot_surface(X,Y,Z,cmap="viridis",alpha=0.45,linewidth=0)
ax.set_zlim(-1.5,1.2)
ax.set_xlabel("X"); ax.set_ylabel("Y"); ax.set_zlabel("Z")
plt.show()
```

## 13. Color Normalization

```python
from matplotlib.colors import Normalize, LogNorm, TwoSlopeNorm

linear = Normalize(vmin=0, vmax=100)       # Linear mapping
logarithmic = LogNorm(vmin=1, vmax=1000)    # Log mapping; values must be positive
centered = TwoSlopeNorm(vmin=-10, vcenter=0, vmax=20)  # Custom center

print(linear([0,25,50,75,100]))
print(logarithmic([1,10,100,1000]))
print(centered([-10,0,20]))
```

## 14. Log-Normalized Surface

```python
x=np.linspace(-4,4,100); y=np.linspace(-4,4,100); X,Y=np.meshgrid(x,y)
Z=np.exp(-0.5*(X**2+Y**2))*1000+1  # Positive values for LogNorm
fig=plt.figure(figsize=(9,7)); ax=fig.add_subplot(111,projection="3d")
s=ax.plot_surface(X,Y,Z,cmap="viridis",norm=LogNorm(vmin=Z.min(),vmax=Z.max()),linewidth=0)
fig.colorbar(s,ax=ax,shrink=0.75,pad=0.10,label="Log-scaled Z")
ax.set_xlabel("X"); ax.set_ylabel("Y"); ax.set_zlabel("Z")
plt.show()
```

## 15. Labels, Limits, and Ticks

```python
t=np.linspace(0,10,100)
fig=plt.figure(figsize=(8,6)); ax=fig.add_subplot(111,projection="3d")
ax.plot(np.cos(t),np.sin(t),t)
ax.set_xlabel("X Coordinate",labelpad=10)  # labelpad = label distance
ax.set_ylabel("Y Coordinate",labelpad=10)
ax.set_zlabel("Z Coordinate",labelpad=10)
ax.set_title("3D Coordinate Plot",pad=15)    # pad = title distance
ax.set_xlim(-1.5,1.5); ax.set_ylim(-1.5,1.5); ax.set_zlim(0,10)
ax.set_xticks([-1,0,1]); ax.set_yticks([-1,0,1]); ax.set_zticks([0,2,4,6,8,10])
plt.show()
```

## 16. Camera Control

```python
t=np.linspace(0,10,100)
fig=plt.figure(figsize=(8,6)); ax=fig.add_subplot(111,projection="3d")
ax.plot(np.cos(t),np.sin(t),t)
ax.view_init(
    elev=25,  # Camera elevation
    azim=45   # Camera rotation around the object
)
ax.set_xlabel("X"); ax.set_ylabel("Y"); ax.set_zlabel("Z")
plt.show()
```

## 17. Perspective vs Orthographic

```python
t=np.linspace(0,10,100)
fig=plt.figure(figsize=(10,5))
ax1=fig.add_subplot(1,2,1,projection="3d"); ax2=fig.add_subplot(1,2,2,projection="3d")
ax1.plot(np.cos(t),np.sin(t),t); ax1.set_proj_type("persp"); ax1.set_title("Perspective")
ax2.plot(np.cos(t),np.sin(t),t); ax2.set_proj_type("ortho"); ax2.set_title("Orthographic")
for ax in (ax1,ax2): ax.set_xlabel("X"); ax.set_ylabel("Y"); ax.set_zlabel("Z")
plt.tight_layout(); plt.show()
```

## 18. Aspect Ratio

```python
x=np.linspace(0,10,100); y=np.linspace(0,2,100); X,Y=np.meshgrid(x,y); Z=np.sin(X)*np.cos(Y)
fig=plt.figure(figsize=(9,6)); ax=fig.add_subplot(111,projection="3d")
ax.plot_surface(X,Y,Z,cmap="viridis",linewidth=0)
ax.set_box_aspect((10,2,2))  # Relative X/Y/Z box proportions
ax.set_xlabel("X"); ax.set_ylabel("Y"); ax.set_zlabel("Z")
plt.show()
```

## 19. 3D Subplots and Mixed 2D/3D

```python
fig=plt.figure(figsize=(12,7))
ax1=fig.add_subplot(2,2,1)
ax2=fig.add_subplot(2,2,2,projection="3d")
ax3=fig.add_subplot(2,1,2)
x=np.linspace(-3,3,100)
ax1.plot(x,np.sin(x)); ax1.set_title("2D Line")
X,Y=np.meshgrid(x,x); Z=np.sin(np.sqrt(X**2+Y**2))
ax2.plot_surface(X,Y,Z,cmap="viridis",linewidth=0); ax2.set_title("3D Surface")
ax3.plot(x,np.cos(x)); ax3.set_title("2D Reference")
plt.tight_layout(); plt.show()
```

## 20. 3D Annotations and Reference Planes

```python
t=np.linspace(0,10,100); x=np.cos(t); y=np.sin(t); z=t
fig=plt.figure(figsize=(9,7)); ax=fig.add_subplot(111,projection="3d")
ax.plot(x,y,z)
ax.text(x[-1],y[-1],z[-1],"End")  # Text in X/Y/Z data coordinates
ax.text2D(0.05,0.95,"Axes annotation",transform=ax.transAxes)  # Axes-relative coordinates
X,Y=np.meshgrid(np.linspace(-2,2,20),np.linspace(-2,2,20)); Z0=np.zeros_like(X)
ax.plot_surface(X,Y,Z0,alpha=0.2,linewidth=0)  # Reference plane at Z=0
ax.set_xlabel("X"); ax.set_ylabel("Y"); ax.set_zlabel("Z")
plt.show()
```

## 21. 3D Vector Fields

```python
x=np.linspace(-2,2,5); y=np.linspace(-2,2,5); z=np.linspace(-2,2,5)
X,Y,Z=np.meshgrid(x,y,z)
U=-Y; V=X; W=np.zeros_like(Z)

fig=plt.figure(figsize=(9,7)); ax=fig.add_subplot(111,projection="3d")
ax.quiver(
    X,Y,Z,U,V,W,
    length=0.25,             # Arrow length scaling
    normalize=True,          # Normalize vector magnitudes
    arrow_length_ratio=0.25  # Relative arrow-head size
)
ax.set_xlabel("X"); ax.set_ylabel("Y"); ax.set_zlabel("Z")
plt.show()
```

### `quiver()` parameters

`X,Y,Z` = arrow positions; `U,V,W` = vector components; `length` = arrow scaling; `normalize` = normalize vector lengths; `arrow_length_ratio` = arrow-head size; `linewidth` and `color` are optional styling parameters.

## 22. Voxels

```python
n=16
coords=np.indices((n,n,n))
cx=cy=cz=(n-1)/2; radius=5
filled=((coords[0]-cx)**2+(coords[1]-cy)**2+(coords[2]-cz)**2 <= radius**2)

fig=plt.figure(figsize=(8,7)); ax=fig.add_subplot(111,projection="3d")
ax.voxels(
    filled,
    alpha=0.75  # Transparency
)
ax.set_xlabel("X"); ax.set_ylabel("Y"); ax.set_zlabel("Z")
ax.set_title("Small 3D Voxel Volume")
plt.show()
```

A voxel is the 3D equivalent of a pixel. `ax.voxels()` is useful for small boolean volumes. Important parameters include `filled`, `facecolors`, `edgecolor`, and `alpha`. Large MRI volumes are usually better handled with dedicated tools such as 3D Slicer, Napari, PyVista, or VTK.

## 23. Synthetic Tumor Segmentation Mask

```python
n=24
coords=np.indices((n,n,n))
cx,cy,cz=12,11,13; rx,ry,rz=8,5,6
tumor_mask=(
    ((coords[0]-cx)/rx)**2 +
    ((coords[1]-cy)/ry)**2 +
    ((coords[2]-cz)/rz)**2 <= 1
)
fig=plt.figure(figsize=(8,7)); ax=fig.add_subplot(111,projection="3d")
ax.voxels(tumor_mask,alpha=0.8)
ax.set_xlabel("Voxel X"); ax.set_ylabel("Voxel Y"); ax.set_zlabel("Voxel Z")
ax.set_title("Illustrative 3D Tumor Segmentation Mask")
plt.show()
```

## 24. MRI Volume Concept

A 3D MRI volume can be viewed conceptually as `Width × Height × Slices`, or X × Y × Z. A segmentation mask can share the same spatial dimensions. Matplotlib is useful for small demonstrations; dedicated medical-imaging software is preferable for large real volumes and interactive analysis.

```python
# Synthetic MRI-like volume for educational purposes only.
rng=np.random.default_rng(5)
volume=rng.normal(size=(20,30,30))
slice_index=10
image_slice=volume[slice_index]
plt.figure(figsize=(6,5))
plt.imshow(image_slice,origin="lower")
plt.xlabel("X"); plt.ylabel("Y")
plt.title(f"Synthetic MRI-like Slice {slice_index}")
plt.colorbar(label="Intensity")
plt.show()
```

## 25. Machine Learning Loss Landscape

```python
w1=np.linspace(-3,3,150); w2=np.linspace(-3,3,150)
W1,W2=np.meshgrid(w1,w2)
loss=(W1**2+W2**2)/5 + np.sin(2*W1) + np.cos(2*W2)
fig=plt.figure(figsize=(10,7)); ax=fig.add_subplot(111,projection="3d")
s=ax.plot_surface(W1,W2,loss,cmap="viridis",rcount=90,ccount=90,linewidth=0)
fig.colorbar(s,ax=ax,shrink=0.75,pad=0.10,label="Loss")
ax.set_xlabel("Parameter 1"); ax.set_ylabel("Parameter 2"); ax.set_zlabel("Loss")
ax.set_title("Illustrative ML Loss Landscape")
plt.show()
```

## 28. Common Mistakes

- **Shape mismatch:** X/Y/Z must describe compatible grids.
- **Missing `projection="3d"`:** a normal Axes object is not a 3D Axes.
- **Old notebook variables:** define data near the plot to avoid accidental reuse.
- **Too much data:** reduce resolution and point count.
- **Misleading color:** explain continuous color with a colorbar.
- **Poor camera angle:** inspect important figures from multiple useful views.
- **Unnecessary 3D:** if 2D is clearer, use 2D.

## 33. Golden Rules

1. Always know what X, Y, and Z represent.
2. Check array shapes before plotting.
3. Use `meshgrid()` for regular surfaces.
4. Use `plot_trisurf()` for irregular samples.
5. Use a colorbar when color represents continuous data.
6. Use legends for categories.
7. Choose camera angles deliberately.
8. Check aspect ratio.
9. Downsample large data.
10. Do not use 3D just for visual effect.
11. Remember that perspective can make comparisons harder.
12. Use specialized tools for large medical volumes.
13. Keep scientific meaning more important than decoration.
14. Define variables near plotting code for reproducibility.

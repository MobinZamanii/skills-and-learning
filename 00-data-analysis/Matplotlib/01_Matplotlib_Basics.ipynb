## 1. What is Matplotlib?

**Matplotlib** is a Python library for creating static, animated, and interactive visualizations.

It is widely used in Data Science, Machine Learning, scientific computing, statistics, Bioinformatics, and research.

The most commonly used interface is `matplotlib.pyplot`, usually imported as `plt`.

## 5. Anatomy of a Matplotlib Figure

```text
Figure
└── Axes
    ├── Title
    ├── X-axis
    ├── Y-axis
    ├── Ticks
    ├── Tick labels
    ├── Legend
    └── Grid
```

- **Figure**: the complete canvas.
- **Axes**: the actual plotting area.
- **Axis**: the x-axis or y-axis.

## 6. Figure Size

````python
plt.figure(figsize=(8, 5))

plt.plot(x, y)
plt.title("Basic Line Plot")
plt.show()
````

## 7. Line Plot

The basic function is `plt.plot(x, y)`.

Important parameters:

- `color`
- `linestyle`
- `linewidth`
- `marker`
- `markersize`
- `markerfacecolor`
- `markeredgecolor`
- `alpha`
- `label`

````python
plt.figure(figsize=(8, 5))

plt.plot(
    x,
    y,
    color="blue",
    linestyle="--",
    linewidth=2,
    marker="o",
    markersize=7,
    markerfacecolor="white",
    markeredgecolor="black",
    alpha=0.9,
    label="Experiment"
)

plt.legend()
plt.show()
````

### 7.1 Line Styles

| Style | Meaning |
|---|---|
| `"-"` | Solid |
| `"--"` | Dashed |
| `"-."` | Dash-dot |
| `":"` | Dotted |

````python
styles = ["-", "--", "-.", ":"]

plt.figure(figsize=(8, 6))

for i, style in enumerate(styles, start=1):
    plt.plot([1, 2, 3, 4], np.array([1, 2, 3, 4]) * i,
             linestyle=style, label=style)

plt.legend()
plt.title("Common Line Styles")
plt.show()
````

### 7.2 Markers

Common markers include `o`, `s`, `^`, `v`, `*`, `+`, `x`, and `D`.

````python
markers = ["o", "s", "^", "v", "*", "+", "x", "D"]

plt.figure(figsize=(10, 6))

for i, marker in enumerate(markers):
    plt.plot(i, i, marker=marker, markersize=10, linestyle="None")

plt.xticks(range(len(markers)), markers)
plt.title("Common Matplotlib Markers")
plt.show()
````

## 8. Titles and Axis Labels

Useful parameters include `fontsize`, `fontweight`, `fontstyle`, `loc`, and `pad`.

````python
plt.figure(figsize=(8, 5))

plt.plot(x, y)

plt.title(
    "Patient Survival",
    fontsize=16,
    fontweight="bold",
    loc="center",
    pad=15
)

plt.xlabel("Age", fontsize=12)
plt.ylabel("Survival Time (months)", fontsize=12)

plt.show()
````

## 9. Legend

Use `label=` inside plotting functions and then call `plt.legend()`.

Important parameters:

- `loc`
- `fontsize`
- `title`
- `frameon`
- `shadow`
- `ncol`

````python
x = np.arange(1, 6)
control = [10, 12, 13, 15, 16]
treatment = [9, 14, 17, 20, 24]

plt.plot(x, control, marker="o", label="Control")
plt.plot(x, treatment, marker="s", label="Treatment")

plt.legend(
    loc="upper left",
    fontsize=10,
    title="Groups",
    frameon=True
)

plt.show()
````

### Common Legend Locations

`upper right`, `upper left`, `lower right`, `lower left`, `center`, `center left`, `center right`, `upper center`, `lower center`.

## 10. Grid

````python
plt.plot(x, treatment, marker="o")

plt.grid(
    visible=True,
    axis="both",
    which="major",
    linestyle="--",
    linewidth=0.5,
    alpha=0.7
)

plt.show()
````

## 11. Axis Limits

````python
plt.plot(x, treatment)

plt.xlim(0, 6)
plt.ylim(0, 30)

plt.show()
````

````python
plt.figure(figsize=(9, 3))

plt.subplot(1, 3, 1)
plt.plot(x, treatment)
plt.axis("tight")
plt.title("tight")

plt.subplot(1, 3, 2)
plt.plot(x, treatment)
plt.axis("equal")
plt.title("equal")

plt.subplot(1, 3, 3)
plt.plot(x, treatment)
plt.axis("off")
plt.title("off")

plt.tight_layout()
plt.show()
````

## 12. Ticks

Use `plt.xticks()` and `plt.yticks()` to control tick positions and labels.

Useful options include custom labels, `rotation`, and `fontsize`.

````python
categories = ["Control", "Treatment", "Follow-up"]
values = [20, 35, 28]

plt.bar(range(len(categories)), values)

plt.xticks(
    range(len(categories)),
    categories,
    rotation=20,
    fontsize=10
)

plt.yticks(fontsize=10)
plt.show()
````

## 13. Multiple Lines

````python
x = np.arange(1, 6)

plt.plot(x, [10, 12, 15, 17, 20], marker="o", label="Gene A")
plt.plot(x, [8, 11, 14, 18, 22], marker="s", label="Gene B")
plt.plot(x, [12, 13, 13, 15, 16], marker="^", label="Gene C")

plt.xlabel("Sample Group")
plt.ylabel("Expression")
plt.title("Gene Expression")
plt.legend()
plt.grid(alpha=0.3)

plt.show()
````

## 14. Shorthand Plot Format

Matplotlib supports compact format strings such as `"r--"`.

- `r` → red
- `--` → dashed

Explicit keyword arguments are usually clearer in production and educational code.

````python
plt.plot(x, treatment, "r--")
plt.show()
````

## 15. Scatter Plot

Use `plt.scatter()` when each observation is represented by a point.

Important parameters:

- `s`
- `c`
- `marker`
- `alpha`
- `edgecolors`
- `linewidths`
- `label`
- `cmap`

````python
age = np.array([25, 31, 42, 50, 58, 65])
survival = np.array([48, 42, 35, 30, 22, 15])

plt.scatter(
    age,
    survival,
    s=80,
    alpha=0.7,
    marker="o",
    edgecolors="black"
)

plt.xlabel("Age")
plt.ylabel("Survival (months)")
plt.title("Age vs Survival")
plt.show()
````

### 15.1 Scatter Plot with Color Mapping

````python
expression = np.array([2.1, 4.5, 3.2, 6.7, 5.1, 7.4])

plt.scatter(
    age,
    survival,
    c=expression,
    cmap="viridis",
    s=100
)

plt.xlabel("Age")
plt.ylabel("Survival (months)")
plt.colorbar(label="Gene Expression")
plt.title("Age vs Survival Colored by Expression")
plt.show()
````

## 16. Bar Chart

Use `plt.bar()` for categorical comparisons.

Important parameters:

- `width`
- `color`
- `edgecolor`
- `linewidth`
- `alpha`
- `align`
- `label`

````python
groups = ["Control", "Treatment A", "Treatment B"]
response = [42, 58, 67]

plt.bar(
    groups,
    response,
    width=0.6,
    edgecolor="black",
    linewidth=1
)

plt.xlabel("Group")
plt.ylabel("Response")
plt.title("Treatment Response")
plt.show()
````

## 17. Horizontal Bar Chart

````python
plt.barh(groups, response, height=0.6)

plt.xlabel("Response")
plt.ylabel("Group")
plt.title("Horizontal Bar Chart")
plt.show()
````

## 18. Grouped Bar Chart

````python
categories = np.array(["A", "B", "C", "D"])
control = np.array([20, 25, 18, 30])
treatment = np.array([28, 31, 24, 35])

x_pos = np.arange(len(categories))
width = 0.35

plt.bar(x_pos - width / 2, control, width, label="Control")
plt.bar(x_pos + width / 2, treatment, width, label="Treatment")

plt.xticks(x_pos, categories)
plt.ylabel("Value")
plt.title("Grouped Bar Chart")
plt.legend()
plt.show()
````

## 19. Histogram

A histogram shows the distribution of numerical data.

Important parameters:

- `bins`
- `range`
- `density`
- `alpha`
- `histtype`
- `edgecolor`

````python
rng = np.random.default_rng(42)
ages = rng.normal(loc=50, scale=10, size=300)

plt.hist(
    ages,
    bins=15,
    edgecolor="black",
    alpha=0.8
)

plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")
plt.show()
````

### Histogram vs Bar Chart

**Bar chart**
- Usually categorical
- Bars represent separate categories

**Histogram**
- Numerical continuous data
- Values are grouped into bins
- Adjacent bins represent ranges

## 20. Pie Chart

Use `plt.pie()` for simple part-to-whole compositions.

Important parameters:

- `labels`
- `autopct`
- `startangle`
- `explode`
- `shadow`
- `counterclock`
- `wedgeprops`

````python
labels = ["Control", "Treatment A", "Treatment B"]
values = [40, 35, 25]

plt.pie(
    values,
    labels=labels,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Sample Composition")
plt.show()
````

````python
explode = [0, 0.08, 0]

plt.pie(
    values,
    labels=labels,
    autopct="%1.1f%%",
    startangle=90,
    explode=explode,
    shadow=False,
    wedgeprops={"edgecolor": "white"}
)

plt.title("Highlighted Treatment Group")
plt.show()
````

## 21. Stack Plot

````python
x = np.arange(1, 6)
group_a = [10, 12, 14, 16, 18]
group_b = [5, 7, 8, 10, 12]
group_c = [3, 4, 5, 6, 7]

plt.stackplot(
    x, group_a, group_b, group_c,
    labels=["Group A", "Group B", "Group C"]
)

plt.xlabel("Time")
plt.ylabel("Value")
plt.title("Stack Plot")
plt.legend(loc="upper left")
plt.show()
````

## 22. Filled Areas with `fill_between()`

`fill_between()` is useful for confidence intervals, uncertainty regions, ranges, and scientific measurements.

````python
x = np.linspace(0, 10, 100)
mean = np.sin(x)
lower = mean - 0.2
upper = mean + 0.2

plt.plot(x, mean, label="Mean")
plt.fill_between(x, lower, upper, alpha=0.25, label="Uncertainty")

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Mean with Uncertainty Region")
plt.legend()
plt.show()
````

## 23. Error Bars

Use `plt.errorbar()` when measurements have uncertainty or standard deviation.

Important parameters:

- `xerr`
- `yerr`
- `fmt`
- `ecolor`
- `capsize`
- `capthick`
- `elinewidth`

````python
groups = np.arange(1, 5)
mean = np.array([20, 25, 23, 30])
std = np.array([2, 3, 1.5, 2.5])

plt.errorbar(
    groups,
    mean,
    yerr=std,
    fmt="o-",
    capsize=5,
    label="Mean ± SD"
)

plt.xlabel("Group")
plt.ylabel("Measurement")
plt.title("Error Bar Plot")
plt.legend()
plt.show()
````

## 24. Logarithmic Scales

````python
x = np.logspace(0, 4, 100)
y = x ** 2

plt.plot(x, y)
plt.xscale("log")
plt.yscale("log")

plt.xlabel("X (log scale)")
plt.ylabel("Y (log scale)")
plt.title("Log-Log Plot")
plt.show()
````

## 25. Reference Lines

````python
x = np.arange(1, 11)
y = np.array([12, 15, 17, 20, 18, 22, 25, 24, 28, 30])

plt.plot(x, y, marker="o")

plt.axhline(y=20, linestyle="--", label="Threshold")
plt.axvline(x=5, linestyle=":", label="Time Point")
plt.axline((1, 12), slope=2, linestyle="-.", label="Reference Trend")

plt.xlabel("Time")
plt.ylabel("Value")
plt.title("Reference Lines")
plt.legend()
plt.show()
````

## 26. Shaded Regions

````python
x = np.arange(1, 11)
y = np.array([12, 15, 17, 20, 18, 22, 25, 24, 28, 30])

plt.plot(x, y)

plt.axhspan(15, 25, alpha=0.15)
plt.axvspan(4, 7, alpha=0.12)

plt.xlabel("Time")
plt.ylabel("Value")
plt.title("Highlighted Regions")
plt.show()
````

## 27. Text and Annotations

Use `plt.text()` for placing text and `plt.annotate()` for pointing to a specific location.

Important annotation parameters include `xy`, `xytext`, `arrowprops`, `fontsize`, `ha`, and `va`.

````python
x = np.arange(1, 6)
y = np.array([10, 14, 18, 25, 22])

plt.plot(x, y, marker="o")

plt.text(3, 18, "Important point", fontsize=11)

plt.annotate(
    "Maximum",
    xy=(4, 25),
    xytext=(3, 30),
    arrowprops={"arrowstyle": "->"},
    ha="center"
)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Text and Annotation")
plt.show()
````

## 32. Subplots

````python
fig, axes = plt.subplots(2, 2, figsize=(10, 7))

axes[0, 0].plot(x, y)
axes[0, 0].set_title("Line Plot")

axes[0, 1].scatter(age, survival)
axes[0, 1].set_title("Scatter Plot")

axes[1, 0].bar(groups, mean)
axes[1, 0].set_title("Bar Chart")

axes[1, 1].hist(ages, bins=12)
axes[1, 1].set_title("Histogram")

plt.tight_layout()
plt.show()
````

### Shared Axes

````python
time = np.arange(1, 5)

control_values = [20, 25, 18, 30]
treatment_values = [28, 31, 24, 35]

fig, axes = plt.subplots(2, 1, figsize=(8, 6), sharex=True)

axes[0].plot(time, control_values, marker="o")
axes[0].set_ylabel("Control")

axes[1].plot(time, treatment_values, marker="s")
axes[1].set_ylabel("Treatment")
axes[1].set_xlabel("Time")

plt.tight_layout()
plt.show()
````

## 33. `tight_layout()` and `subplots_adjust()`

````python
fig, axes = plt.subplots(1, 2, figsize=(9, 4))

for ax in axes:
    ax.plot(x, treatment)
    ax.set_xlabel("X")
    ax.set_ylabel("Y")

plt.subplots_adjust(wspace=0.4)
plt.show()
````

## 40. Scientific Example — Patient Survival

````python
patients = pd.DataFrame({
    "Age": [35, 42, 50, 58, 63, 70],
    "Survival_Months": [48, 42, 36, 29, 24, 18],
    "Group": ["A", "A", "B", "B", "B", "A"]
})

plt.figure(figsize=(8, 5))

for group in patients["Group"].unique():
    subset = patients[patients["Group"] == group]
    plt.scatter(
        subset["Age"],
        subset["Survival_Months"],
        s=90,
        label=f"Group {group}"
    )

plt.xlabel("Age (years)")
plt.ylabel("Survival (months)")
plt.title("Patient Age vs Survival")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
````

## 44. Choosing the Right Plot

| Goal | Recommended plot |
|---|---|
| Trend over ordered values | Line plot |
| Relationship between two numeric variables | Scatter plot |
| Compare categories | Bar chart |
| Distribution of numerical data | Histogram |
| Part-to-whole composition | Pie chart, when appropriate |
| Uncertainty around measurements | Error bars / filled region |
| Multiple components over time | Stack plot |

## 45. Common Mistakes

1. Mismatched X and Y lengths.
2. Missing or unclear axis labels.
3. Too much decoration.
4. Choosing the wrong chart type.
5. Saving important figures at insufficient resolution.
6. Confusing `color` and `c`.
7. Using only pyplot-style commands for complex figures.
8. Using a log scale without understanding the data.

A good visualization should prioritize **clarity, accuracy, and scientific meaning**.

# 51. Golden Rules

1. **Always label important axes.**
2. **Use meaningful titles.**
3. **Choose the plot type based on the question.**
4. **Keep figures readable and uncluttered.**
5. **Use legends when multiple groups are present.**
6. **Use transparency when overlapping elements need to remain visible.**
7. **Use log scales only when they make analytical sense.**
8. **Show uncertainty when uncertainty matters.**
9. **Save research figures at an appropriate resolution.**
10. **Prefer the Object-Oriented API for complex figures.**
11. **Do not sacrifice scientific clarity for decoration.**
12. **Make every figure understandable without hidden context.**
